````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
DB_PATH = "midterm_a_events.db"
ACTIVITIES = ["迎新晚會", "程式工作坊", "籃球友誼賽"]
ALL_ACTIVITIES = "全部活動"
ACTIONS = ["報名", "取消", "查詢"]

def init_db():
    conn = sqlite3.connect(DB_PATH)
    try:
        conn.execute("PRAGMA foreign_keys = ON")
        conn.execute("DROP TABLE IF EXISTS registrations")
        conn.execute("DROP TABLE IF EXISTS activities")
        conn.execute("""CREATE TABLE activities (
            id INTEGER PRIMARY KEY,
            name TEXT UNIQUE NOT NULL,
            event_date TEXT NOT NULL,
            capacity INTEGER NOT NULL CHECK(capacity >= 0)
        )""")
        conn.execute("""CREATE TABLE registrations (
            id INTEGER PRIMARY KEY,
            activity_id INTEGER NOT NULL,
            student_name TEXT NOT NULL,
            attendees INTEGER NOT NULL CHECK(attendees BETWEEN 1 AND 3),
            status TEXT NOT NULL CHECK(status IN ('已報名', '已取消')),
            created_at TEXT DEFAULT CURRENT_TIMESTAMP,
            FOREIGN KEY (activity_id) REFERENCES activities(id),
            UNIQUE(activity_id, student_name)
        )""")
        conn.executemany(
            "INSERT INTO activities (name, event_date, capacity) VALUES (?, ?, ?)",
            [("迎新晚會", "2026-11-20", 6),
             ("程式工作坊", "2026-11-25", 4),
             ("籃球友誼賽", "2026-11-28", 3)]
        )
        conn.commit()
        print("✅ 已重建 2 張資料表並載入 3 場活動")
    finally:
        conn.close()

init_db()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
AI_CACHE = {}  # 相同句子只呼叫 Groq 一次，節省 API 額度

SYSTEM_PROMPT = """你是系學會活動報名系統的語意分析器。
你的任務只有從使用者的一句話擷取三個欄位，必須僅回傳一行：動作,活動,人數
動作只能是：報名、取消、查詢、未知。
活動只能是：迎新晚會、程式工作坊、籃球友誼賽、全部活動、未知。
使用者要求各活動報名狀況、所有活動統計、整體報名統計等，請回傳：查詢,全部活動,1。
報名可包含「參加」「幫我留名額」；取消可包含「退報」「不參加了」；查詢可包含「剩餘名額」「還有幾位」。
人數須輸出阿拉伯數字；未說人數一律 1。
「我和兩個朋友」是 3；「我和一位朋友」是 2；「四個人」是 4。
不允許辨識其他活動。若動作或活動無法確定，回傳：未知,未知,1。
不可回傳說明、Markdown、JSON、姓名、SQL，或其他內容。
例如：我和兩個朋友參加籃球友誼賽 -> 報名,籃球友誼賽,3
例如：各活動報名狀況 -> 查詢,全部活動,1
例如：我要報名歌唱比賽 -> 未知,未知,1
"""

def parse_ai_reply(raw):
    # 僅讀取第一行及三個欄位，不直接信任 Groq 的答案
    first_line = str(raw).strip().splitlines()[0] if str(raw).strip() else ""
    first_line = first_line.strip("` 	")
    parts = re.split(r"[,，、|｜]", first_line)
    if len(parts) < 3:
        return ("未知", "未知", 1)
    action_text, event_text, count_text = [p.strip() for p in parts[:3]]
    action = action_text if action_text in ACTIONS else "未知"
    valid_events = ACTIVITIES + [ALL_ACTIVITIES]
    event = event_text if event_text in valid_events else "未知"
    found = re.search(r"-?\d+", count_text)
    count = int(found.group(0)) if found else 1
    # 只有查詢可以用「全部活動」；無效輸出一律退回未知
    if action == "未知" or event == "未知" or (event == ALL_ACTIVITIES and action != "查詢"):
        return ("未知", "未知", count)
    return (action, event, count)

def ai_interpret(user_text):
    text = str(user_text or "").strip()
    if not text:
        return ("未知", "未知", 1, "（空白輸入）")
    if text in AI_CACHE:
        return AI_CACHE[text]

    raw = ""
    for attempt in range(3):  # 總共最多 3 次呼叫
        try:
            response = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": text},
                ],
            )
            raw = (response.choices[0].message.content or "").strip()
            break
        except Exception as exc:
            error_text = str(exc)
            if "429" in error_text and attempt < 2:
                print(f"⚠️ Groq 額度限制，30 秒後重試（第 {attempt + 1} 次）")
                time.sleep(30)
            else:
                print("⚠️ AI 解讀失敗，將以未知處理（請檢查 Groq 設定／額度）")
                break
        finally:
            time.sleep(2)  # 每次實際 API 呼叫後間隔 2 秒

    action, event, count = parse_ai_reply(raw)
    result = (action, event, count, raw)
    AI_CACHE[text] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
AI_TESTS = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),
    ("我和兩個朋友想參加籃球友誼賽", "報名", "籃球友誼賽", 3),
    ("幫我取消程式工作坊", "取消", "程式工作坊", 1),
    ("程式工作坊還有幾個名額", "查詢", "程式工作坊", 1),
    ("各活動報名狀況", "查詢", "全部活動", 1),
    ("我要報名迎新晚會三人", "報名", "迎新晚會", 3),
    ("我要報名迎新晚會四個人", "報名", "迎新晚會", 4),
    ("我要報名歌唱比賽", "未知", "未知", "*"),
    ("今天天氣如何", "未知", "未知", "*"),
]

ai_passed = 0
for idx, (sentence, exp_action, exp_event, exp_count) in enumerate(AI_TESTS, 1):
    action, event, count, raw = ai_interpret(sentence)
    passed = (action == exp_action and event == exp_event and
              (exp_count == "*" or count == exp_count))
    ai_passed += int(passed)
    print(f"[{idx}] {'✅ PASS' if passed else '❌ FAIL'}：{sentence}")
    print("  AI 原始回覆：", raw)
    print("  系統解讀：", (action, event, count), "｜預期：", (exp_action, exp_event, exp_count))
print(f"\nAI 解讀測試通過 {ai_passed}/{len(AI_TESTS)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def _get_activity(conn, event_name):
    return conn.execute(
        "SELECT id, name, event_date, capacity FROM activities WHERE name = ?",
        (event_name,)
    ).fetchone()

def _occupied(conn, activity_id):
    row = conn.execute(
        "SELECT COALESCE(SUM(attendees), 0) FROM registrations WHERE activity_id = ? AND status = ?",
        (activity_id, "已報名")
    ).fetchone()
    return int(row[0]) if row is not None else 0

def register_event(name, event_name, count):
    name = str(name or "").strip()
    if not name:
        return "❌ 報名需要填寫姓名。"
    if not isinstance(count, int) or not 1 <= count <= 3:
        return "❌ 報名人數只能介於 1～3 人（含本人）。"
    conn = sqlite3.connect(DB_PATH, timeout=10)
    try:
        conn.execute("PRAGMA foreign_keys = ON")
        conn.execute("BEGIN IMMEDIATE")  # 在同一交易內核對與寫入，避免超額
        activity = _get_activity(conn, event_name)
        if activity is None:
            return "❌ 查無此活動。"
        activity_id, _, _, capacity = activity
        existing = conn.execute(
            "SELECT id, status FROM registrations WHERE activity_id = ? AND student_name = ?",
            (activity_id, name)
        ).fetchone()
        if existing is not None and existing[1] == "已報名":
            return f"❌ {name} 已經報名「{event_name}」，不可重複報名。"
        active_count = conn.execute(
            "SELECT COUNT(*) FROM registrations WHERE student_name = ? AND status = ?",
            (name, "已報名")
        ).fetchone()
        if active_count is not None and active_count[0] >= 2:
            return f"❌ {name} 已同時報名 2 場活動，不能再新增。"
        remaining = capacity - _occupied(conn, activity_id)
        if remaining < count:
            return f"❌ 「{event_name}」剩餘 {remaining} 人名額，不足 {count} 人報名。"
        if existing is None:
            conn.execute(
                "INSERT INTO registrations (activity_id, student_name, attendees, status) VALUES (?, ?, ?, ?)",
                (activity_id, name, count, "已報名")
            )
        else:
            conn.execute(
                "UPDATE registrations SET attendees = ?, status = ? WHERE id = ?",
                (count, "已報名", existing[0])
            )
        conn.commit()
        return f"✅ {name} 成功報名「{event_name}」共 {count} 人；剩餘 {remaining-count} 人名額。"
    except sqlite3.Error:
        conn.rollback()
        return "❌ 資料庫處理失敗，請稍後重試。"
    finally:
        conn.close()

def cancel_event(name, event_name):
    name = str(name or "").strip()
    if not name:
        return "❌ 取消報名需要填寫姓名。"
    conn = sqlite3.connect(DB_PATH, timeout=10)
    try:
        conn.execute("BEGIN IMMEDIATE")
        activity = _get_activity(conn, event_name)
        if activity is None:
            return "❌ 查無此活動。"
        activity_id = activity[0]
        record = conn.execute(
            "SELECT id, attendees FROM registrations WHERE activity_id = ? AND student_name = ? AND status = ?",
            (activity_id, name, "已報名")
        ).fetchone()
        if record is None:
            return f"❌ 查不到 {name} 在「{event_name}」的有效報名紀錄。"
        conn.execute("UPDATE registrations SET status = ? WHERE id = ?", ("已取消", record[0]))
        # 在同一個交易內重新計算名額，讓回應與規格書的預期值一致
        remaining = activity[3] - _occupied(conn, activity_id)
        conn.commit()
        return (f"✅ 已取消 {name} 的「{event_name}」報名，"
                f"釋出 {record[1]} 人名額；剩餘 {remaining} 人名額。")
    except sqlite3.Error:
        conn.rollback()
        return "❌ 資料庫處理失敗，請稍後重試。"
    finally:
        conn.close()

def query_event(event_name):
    conn = sqlite3.connect(DB_PATH)
    try:
        activity = _get_activity(conn, event_name)
        if activity is None:
            return "❌ 查無此活動。"
        activity_id, _, event_date, capacity = activity
        occupied = _occupied(conn, activity_id)
        return (f"📅 {event_name}（{event_date}）：名額 {capacity} 人，"
                f"已報名 {occupied} 人，剩餘 {capacity - occupied} 人。")
    finally:
        conn.close()

def report_all_events():
    conn = sqlite3.connect(DB_PATH)
    try:
        # LEFT JOIN 保留沒人報名的活動；GROUP BY 分組計算實際人數
        rows = conn.execute("""
            SELECT a.name, a.event_date, a.capacity,
                   COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.attendees ELSE 0 END), 0) AS occupied
            FROM activities AS a
            LEFT JOIN registrations AS r ON r.activity_id = a.id
            GROUP BY a.id, a.name, a.event_date, a.capacity
            ORDER BY a.id
        """).fetchall()
        lines = ["📊 各活動報名統計（人數以含攜伴總人數計）"]
        for name, event_date, capacity, occupied in rows:
            lines.append(f"{name}｜{event_date}｜名額 {capacity}｜已報名 {occupied}｜剩餘 {capacity-occupied}")
        return "\n".join(lines)
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def process_request(user_text, student_name=""):
    action, event, count, raw = ai_interpret(user_text)
    if action == "未知" or event == "未知":
        return "⚠️ 無法判斷需求。目前活動：" + "、".join(ACTIVITIES) + "。請重新輸入。"
    if action == "查詢":
        return report_all_events() if event == ALL_ACTIVITIES else query_event(event)
    if event == ALL_ACTIVITIES:
        return "❌ 報名或取消必須指定單一活動。"
    if action == "報名":
        return register_event(student_name, event, count)
    if action == "取消":
        return cancel_event(student_name, event)
    return "⚠️ 無法判斷需求。目前活動：" + "、".join(ACTIVITIES)

# 每步都核對「回覆文字」以及「資料庫實際占用人數／有效報名筆數」
# 每個元組依序為：迎新晚會、程式工作坊、籃球友誼賽
# 與規格書第 7-2 節相同，恰好 10 步。
FLOW_TESTS = [
    ("小明", "我和兩個朋友想參加迎新晚會", "成功報名", (3, 0, 0), (1, 0, 0)),
    ("小明", "我要報名迎新晚會", "不可重複報名", (3, 0, 0), (1, 0, 0)),
    ("小華", "我要報名程式工作坊", "成功報名", (3, 1, 0), (1, 1, 0)),
    ("小華", "我要報名迎新晚會", "成功報名", (4, 1, 0), (2, 1, 0)),
    ("小華", "我要報名籃球友誼賽", "已同時報名 2 場", (4, 1, 0), (2, 1, 0)),
    ("小美", "我和兩個朋友想參加籃球友誼賽", "成功報名", (4, 1, 3), (2, 1, 1)),
    ("小強", "我要報名籃球友誼賽", "剩餘 0 人名額", (4, 1, 3), (2, 1, 1)),
    ("小明", "幫我取消迎新晚會", "剩餘 5 人名額", (1, 1, 3), (1, 1, 1)),
    ("", "籃球友誼賽還有幾個名額", "剩餘 0 人", (1, 1, 3), (1, 1, 1)),
    ("", "各活動報名狀況", "各活動報名統計", (1, 1, 3), (1, 1, 1)),
]

def flow_db_snapshot():
    """實際從 SQLite 計算有效報名占用人數及報名紀錄數。"""
    conn = sqlite3.connect(DB_PATH)
    try:
        rows = conn.execute("""
            SELECT a.id,
                   COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.attendees ELSE 0 END), 0),
                   COALESCE(SUM(CASE WHEN r.status = '已報名' THEN 1 ELSE 0 END), 0)
            FROM activities AS a
            LEFT JOIN registrations AS r ON r.activity_id = a.id
            GROUP BY a.id ORDER BY a.id
        """).fetchall()
        return (tuple(int(r[1]) for r in rows), tuple(int(r[2]) for r in rows))
    finally:
        conn.close()

flow_passed = 0
for idx, (name, sentence, expected_piece, expected_occupied, expected_active) in enumerate(FLOW_TESTS, 1):
    result = process_request(sentence, name)
    occupied, active = flow_db_snapshot()
    passed = (expected_piece in result and occupied == expected_occupied and active == expected_active)
    if idx == 8:
        # 額外檢查「取消」是修改狀態，不是刪掉歷史紀錄
        conn = sqlite3.connect(DB_PATH)
        try:
            cancelled = conn.execute("""
                SELECT r.status, r.attendees FROM registrations AS r
                JOIN activities AS a ON a.id = r.activity_id
                WHERE a.name = ? AND r.student_name = ?
            """, ("迎新晚會", "小明")).fetchone()
            passed = passed and cancelled == ("已取消", 3)
        finally:
            conn.close()
    if idx == 10:
        # 確認回覆包含三場活動的日期、名額及正確統計
        required_lines = [
            "迎新晚會｜2026-11-20｜名額 6｜已報名 1｜剩餘 5",
            "程式工作坊｜2026-11-25｜名額 4｜已報名 1｜剩餘 3",
            "籃球友誼賽｜2026-11-28｜名額 3｜已報名 3｜剩餘 0",
        ]
        passed = passed and all(line in result for line in required_lines)
    flow_passed += int(passed)
    print(f"[{idx}] {'✅ PASS' if passed else '❌ FAIL'}｜姓名：{name or '（不需要）'}｜{sentence}")
    print("  回應：", result)
    print(f"  DB 占用：{occupied}（預期 {expected_occupied}）｜有效筆數：{active}（預期 {expected_active}）")
print(f"\n流程測試通過 {flow_passed}/{len(FLOW_TESTS)} 步")
show_tables()
````
